# 02 · Series and DataFrame Creation & Attributes

**Goal:** learn every common way to build `Series` and `DataFrame` objects, and how to
inspect their shape, columns, index, and dtypes.

### Creating a `Series`

```python
import pandas as pd
import numpy as np

# From a list -- gets a default integer index (0, 1, 2, ...)
s1 = pd.Series([10, 20, 30, 40])
print(s1)

# From a list WITH a custom index
s2 = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s2)

# From a dictionary -- keys automatically become the index
s3 = pd.Series({"NYC": 8_400_000, "LA": 3_900_000, "Chicago": 2_700_000})
print(s3)
```

### `Series` attributes

```python
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"], name="values")

print("values:", s.values)   # underlying NumPy array
print("index: ", s.index)
print("name:  ", s.name)
print("dtype: ", s.dtype)
print("shape: ", s.shape)
print("size:  ", s.size)
```

### Creating a `DataFrame`

There are several common ways — from a dict of lists, a list of dicts, a NumPy array, or
another `Series`/`DataFrame`.

```python
# 1. From a dict of lists -- keys become column names (most common)
df1 = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 32, 18],
    "score": [88.5, 92.1, 79.3]
})
print(df1)
```

```python
# 2. From a list of dicts -- each dict is one ROW
df2 = pd.DataFrame([
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 32},
    {"name": "Charlie", "age": 18}
])
print(df2)
```

```python
# 3. From a NumPy array -- columns get default names unless you specify them
arr = np.array([[1, 2, 3], [4, 5, 6]])
df3 = pd.DataFrame(arr, columns=["a", "b", "c"])
print(df3)

# 4. Empty DataFrame you fill in later
df4 = pd.DataFrame(columns=["x", "y"])
print(df4)
print(df4.shape)
```

### Core DataFrame attributes

These are the ones you'll reach for constantly, especially right after loading any dataset.

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "age": [25, 32, 18, 47],
    "city": ["NYC", "LA", "Chicago", "Houston"],
    "score": [88.5, 92.1, 79.3, 85.0]
})

print("shape:  ", df.shape)     # (rows, columns)
print("columns:", df.columns.tolist())
print("index:  ", df.index.tolist())
print("size:   ", df.size)      # total number of cells
print()
print("dtypes:")
print(df.dtypes)
```

### Selecting a single column returns a `Series`

This is a foundational fact: a `DataFrame` is really a collection of `Series` that share an
index, and selecting one column gives you that `Series` back.

```python
ages = df["age"]      # square-bracket column access
print(ages)
print(type(ages))       # Series!

# Selecting MULTIPLE columns (note the double brackets) returns a DataFrame, not a Series
subset = df[["name", "age"]]
print(subset)
print(type(subset))      # DataFrame
```

### Converting a `Series` back to a NumPy array or Python list

```python
ages = df["age"]

print(ages.values)         # NumPy array
print(ages.to_numpy())     # same thing, more explicit/preferred in modern pandas
print(ages.tolist())        # plain Python list
```

### Adding and removing columns

Adding a new column is as simple as assigning to a new key; removing uses `.drop()`.

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 32, 18]
})

df["is_adult"] = df["age"] >= 18       # new column computed from an existing one
print(df)

df["country"] = "USA"                    # new column, same value for every row (broadcast)
print(df)

df_dropped = df.drop(columns=["country"])   # returns a NEW DataFrame by default
print(df_dropped)
print()
print(df)   # original DataFrame is untouched -- drop() doesn't modify in place unless told to
```

### 🧠 Quick check

1. What type of object do you get from `df["age"]` vs. `df[["age"]]`?
2. What becomes the index when you build a `Series` from a dictionary?
3. Does `df.drop(columns=[...])` modify the original DataFrame by default?

<details>
<summary>Answers</summary>

1. `df["age"]` (single brackets, one column name) returns a `Series`; `df[["age"]]` (double
   brackets, a list of column names) returns a `DataFrame` with just that one column.
2. The dictionary's keys become the index; the values become the data.
3. No — by default it returns a new DataFrame, leaving the original unchanged (pass
   `inplace=True` to modify it directly, though returning a new object is usually preferred).
</details>
