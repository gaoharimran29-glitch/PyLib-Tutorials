**Goal of this notebook:** understand *what* pandas is, *why* it exists, and meet the two core

### What is pandas?

Pandas is Python's primary library for working with **labeled, tabular data** — think
spreadsheets or SQL tables, but programmable. It's built on top of NumPy, so it inherits
NumPy's speed, but adds:

- **Labels** for rows and columns (instead of just numeric positions)
- Built-in handling of **missing data**
- Powerful **grouping, filtering, merging, and reshaping** tools
- Easy reading/writing of CSV, Excel, SQL, JSON, and more

If NumPy is about raw numerical arrays, pandas is about **real-world, messy, labeled data**.

### The two core objects: `Series` and `DataFrame`

- **`Series`** — a single labeled column of data (like one spreadsheet column, or a
  dictionary with ordered keys).
- **`DataFrame`** — a full labeled table made of multiple `Series` sharing the same row
  labels (like an entire spreadsheet or SQL table).

```python
import pandas as pd
import numpy as np

# A Series: one labeled column
ages = pd.Series([25, 32, 18, 47], name="age")
print(ages)
print(type(ages))
```

```python
# A DataFrame: a full table (dict of columns, each a Series)
data = {
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "age": [25, 32, 18, 47],
    "city": ["NYC", "LA", "Chicago", "Houston"]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

### Why not just use a NumPy array or a list of dictionaries?

Let's see the difference concretely: pandas keeps column names, mixed types, and gives you a
much richer toolbox for exploring and cleaning data than raw arrays or lists of dicts do.

```python
# NumPy struggles with mixed types (names AND ages AND cities) in one array
try:
    arr = np.array([["Alice", 25, "NYC"], ["Bob", 32, "LA"]])
    print(arr)
    print(arr.dtype)   # everything becomes strings -- ages lose their numeric type!
except Exception as e:
    print(e)

# pandas keeps each column's own appropriate dtype
print("\npandas dtypes:")
print(df.dtypes)
```

### Every DataFrame has a row **index** and column **labels**

This labeling is the core idea that makes pandas so powerful — you select data by meaningful
names, not just numeric position (though numeric position still works too, see notebook 03).

```python
print(df.index)     # the row labels (default: 0, 1, 2, 3...)
print(df.columns)   # the column labels

# You can set a more meaningful index, e.g. using "name" as the row label
df_named = df.set_index("name")
print(df_named)
print(df_named.index)
```

### A first glance at pandas' power: one-line operations

These will each get their own deep-dive notebook, but here's a preview of how expressive
pandas is even in a single line.

```python
print(df["age"].mean())              # average age
print(df[df["age"] > 25])             # filter: rows where age > 25
print(df.sort_values("age"))          # sort by age
print(df.groupby("city")["age"].mean())  # average age per city
```

### Key vocabulary (cheat-sheet)

- **`Series`** — one labeled 1D column of data
- **`DataFrame`** — a labeled 2D table, a collection of `Series` sharing an index
- **Index** — the row labels of a `Series`/`DataFrame`
- **Column** — a named `Series` within a `DataFrame`
- **`dtype`** — the data type of a column (`int64`, `float64`, `object` for strings, etc.)
- **`NaN`** — pandas' marker for a missing value (covered in depth in notebook 04)
