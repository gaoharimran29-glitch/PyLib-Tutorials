### What is Object-Oriented Programming?

Object-Oriented Programming (OOP) is a way of structuring code around **objects** — bundles of
**data** (attributes) and **behavior** (methods) — instead of around a sequence of instructions.

Think of a real-world object, like a car:
- It **has** properties: color, brand, speed, fuel level → these become **attributes**.
- It **can do** things: accelerate, brake, honk → these become **methods**.

In OOP, we model this with a **class** (the blueprint) and **objects** (the actual cars built
from that blueprint).

### Procedural vs Object-Oriented style

Let's see the *same* problem solved two ways, so the motivation for OOP is concrete.

```python
# --- Procedural style: data and functions are separate ---

def create_car(brand, speed):
    return {"brand": brand, "speed": speed}

def accelerate(car, amount):
    car["speed"] += amount
    return car

my_car = create_car("Toyota", 0)
my_car = accelerate(my_car, 20)
print(my_car)

# Problem: nothing stops another part of the program from doing my_car["speed"] = "banana"
# There's no guarantee the dict always represents a valid car.
```

```python
# --- Object-Oriented style: data and behavior live together ---

class Car:
    def __init__(self, brand, speed=0):
        self.brand = brand
        self.speed = speed

    def accelerate(self, amount):
        self.speed += amount

my_car = Car("Toyota")
my_car.accelerate(20)
print(my_car.brand, my_car.speed)

# Now "accelerate" is a behavior that belongs to Car objects.
# Later notebooks show how to also *protect* the data from invalid values (encapsulation).
```

### Why bother with OOP?

| Benefit | What it means |
|---|---|
| **Modularity** | Each class is a self-contained unit you can build/test/reuse independently |
| **Reusability** | Inheritance lets you reuse and extend existing classes |
| **Scalability** | Large codebases stay organized as "things" instead of tangled functions |
| **Maintainability** | Changing how a `Car` works only requires touching the `Car` class |
| **Real-world modeling** | Code maps naturally to real entities: users, orders, accounts, shapes... |

### The Four Pillars of OOP

This tutorial series has one notebook dedicated to each pillar:

1. **Encapsulation** — bundling data + methods, and controlling access to the data
2. **Inheritance** — creating new classes based on existing ones
3. **Polymorphism** — the same method name behaving differently depending on the object
4. **Abstraction** — hiding complex implementation details behind a simple interface

### Key vocabulary (keep this as a cheat-sheet)

- **Class** — a blueprint/template for creating objects (e.g. `Car`)
- **Object / Instance** — a concrete thing built from a class (e.g. `my_car`)
- **Attribute** — a variable that belongs to an object or class (e.g. `speed`)
- **Method** — a function that belongs to a class (e.g. `accelerate`)
- **`self`** — refers to the specific object a method is being called on
- **Constructor** — the special method (`__init__`) that runs when an object is created
