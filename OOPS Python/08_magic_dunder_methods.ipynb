# 08 · Magic Methods (Dunder Methods)

**Goal:** learn the special `__x__` methods that let your custom objects integrate naturally
with Python's built-in syntax — printing, comparisons, arithmetic, containers, and more.

### What are "dunder" methods?

"Dunder" = **D**ouble **UNDER**score. Methods like `__init__`, `__str__`, `__add__` are called
automatically by Python in response to built-in syntax (`print()`, `+`, `==`, `len()`, `[]`,
`for`, etc.), instead of you calling them by name directly. This is how Python achieves
operator overloading and lets custom objects "feel" like built-in types.

### `__str__` vs `__repr__` — string representations

- `__str__` → "informal", readable string, used by `print()` and `str()`. Aimed at humans.
- `__repr__` → "official", unambiguous string, used in the REPL, inside lists, and as a fallback
  if `__str__` isn't defined. Aimed at developers/debugging — ideally something you could
  copy-paste to recreate the object.

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 4)
print(p)          # uses __str__  -> (3, 4)
print(str(p))     # same
print(repr(p))    # uses __repr__ -> Point(x=3, y=4)

points = [p, Point(1, 1)]
print(points)     # lists show elements using __repr__, not __str__
```

### Arithmetic operators

Each operator maps to a dunder method: `+` → `__add__`, `-` → `__sub__`, `*` → `__mul__`, etc.

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(1, 1)

print(v1 + v2)     # calls v1.__add__(v2)
print(v1 - v2)     # calls v1.__sub__(v2)
print(v1 * 3)      # calls v1.__mul__(3)
```

### Comparison operators

`==` → `__eq__`, `<` → `__lt__`, `<=` → `__le__`, `>` → `__gt__`, etc.
Without `__eq__`, Python compares objects by identity (memory address), NOT by their values!

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __eq__(self, other):
        return self.amount == other.amount

    def __lt__(self, other):
        return self.amount < other.amount

    def __repr__(self):
        return f"Money({self.amount})"

m1 = Money(100)
m2 = Money(100)
m3 = Money(50)

print(m1 == m2)   # True -- compares VALUES because we defined __eq__
print(m1 == m3)   # False
print(m3 < m1)    # True -- uses __lt__

# Bonus: once __lt__ exists, sorted() works automatically!
print(sorted([m1, m3, Money(75)]))
```

### Container-like behavior: `__len__`, `__getitem__`, `__contains__`

These let your objects behave like lists/dicts — support `len()`, indexing `obj[i]`,
and the `in` keyword.

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, index):
        return self.songs[index]

    def __contains__(self, song):
        return song in self.songs

p = Playlist(["Song A", "Song B", "Song C"])

print(len(p))            # __len__
print(p[0])               # __getitem__
print("Song B" in p)      # __contains__

for song in p:             # __getitem__ also enables iteration automatically!
    print("-", song)
```

### `__call__` — making an object callable like a function

If a class defines `__call__`, its instances can be "called" with `()` just like a function.

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor

double = Multiplier(2)
triple = Multiplier(3)

print(double(5))    # calls double.__call__(5) -> 10
print(triple(5))    # 15
```

### Quick reference table

| Syntax | Dunder method |
|---|---|
| `print(obj)` / `str(obj)` | `__str__` |
| `repr(obj)` | `__repr__` |
| `obj1 + obj2` | `__add__` |
| `obj1 - obj2` | `__sub__` |
| `obj1 == obj2` | `__eq__` |
| `obj1 < obj2` | `__lt__` |
| `len(obj)` | `__len__` |
| `obj[i]` | `__getitem__` |
| `x in obj` | `__contains__` |
| `obj(...)` | `__call__` |
| `for x in obj` | `__iter__` (or `__getitem__` as fallback) |
