# 03 · Constructors (`__init__`)

**Goal:** understand exactly when and how `__init__` runs, default arguments, and the
difference between `__new__` and `__init__` (briefly).

### What is a constructor?

A constructor is a method that automatically runs **when an object is created**. In Python,
that method is `__init__` (short for "initialize"). It's where you typically set up an
object's starting attributes.

```python
class Point:
    def __init__(self, x, y):
        print("Constructor called!")
        self.x = x
        self.y = y

p = Point(3, 4)     # "Constructor called!" prints automatically
print(p.x, p.y)
```

### Default parameter values

Just like regular functions, `__init__` parameters can have defaults, making them optional.

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

origin = Point()          # uses defaults: (0, 0)
p2 = Point(5, 10)         # overrides both
p3 = Point(x=7)           # overrides only x, y stays 0

print(origin.x, origin.y)
print(p2.x, p2.y)
print(p3.x, p3.y)
```

### Python does NOT support constructor overloading (like Java/C++)

You can't define `__init__` twice with different parameter lists — the second definition
just replaces the first. Instead, use default values, `*args`/`**kwargs`, or classmethods
(see notebook 09) to support multiple "ways" of constructing an object.

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    @classmethod
    def from_tuple(cls, coords):
        # an alternative constructor
        return cls(coords[0], coords[1])

p1 = Point(1, 2)
p2 = Point.from_tuple((5, 6))
print(p1.x, p1.y)
print(p2.x, p2.y)
```

### `__init__` vs `__new__` (good to know, rarely used directly)

- `__new__` actually **creates** the object (allocates memory) — runs first.
- `__init__` **initializes** the already-created object — runs second.

99% of the time you only ever write `__init__`. `__new__` is mostly used for advanced patterns
like singletons or immutable types — not something beginners need to touch, but it explains
*why* `__init__` doesn't need to `return` anything (the object already exists).

```python
class Demo:
    def __new__(cls, *args, **kwargs):
        print("1. __new__ runs first (creates the object)")
        return super().__new__(cls)

    def __init__(self, value):
        print("2. __init__ runs second (initializes the object)")
        self.value = value

d = Demo(42)
```

### Validating input inside the constructor

A common, practical use of `__init__` is validating data as soon as the object is created,
so invalid objects can never exist in the first place.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative")
        self.owner = owner
        self.balance = balance

acc = BankAccount("Asha", 100)
print(acc.owner, acc.balance)

try:
    bad = BankAccount("Ravi", -50)
except ValueError as e:
    print("Error caught:", e)
```
