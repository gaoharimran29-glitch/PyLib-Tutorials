**Goal:** learn the exact syntax for defining a class, creating objects, and understanding

### Defining a class

Use the `class` keyword. By convention, class names use `PascalCase` (capitalize each word).

```python
class Dog:
    pass  # an empty class — valid but not very useful yet

# Creating (instantiating) objects from the class
d1 = Dog()
d2 = Dog()

print(type(d1))
print(d1 is d2)   # False — they are two separate objects in memory
```

### Adding attributes and methods

Attributes are usually set inside `__init__` (covered in depth in notebook 03).
Methods are just functions defined inside the class — their first parameter is always `self`.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name      # instance attribute
        self.breed = breed    # instance attribute

    def bark(self):            # instance method
        return f"{self.name} says Woof!"

d1 = Dog("Rex", "Labrador")
d2 = Dog("Bella", "Poodle")

print(d1.bark())
print(d2.bark())

# Each object keeps its OWN copy of instance attributes:
print(d1.name, d2.name)
```

### Why `self`?

`self` is how a method refers to the *specific object* it was called on. When you write
`d1.bark()`, Python actually does `Dog.bark(d1)` behind the scenes — `self` becomes `d1`.

You don't pass `self` yourself; Python fills it in automatically when calling via an object.

```python
# Proof that d1.bark() and Dog.bark(d1) are the same thing:
print(d1.bark())
print(Dog.bark(d1))
```

### Instance attributes vs. class attributes

- **Instance attributes** belong to one specific object (set via `self.x = ...`, usually in
  `__init__`). Each object has its own copy.
- **Class attributes** belong to the class itself and are **shared** by all instances, unless
  an instance overrides them.

```python
class Dog:
    species = "Canis familiaris"   # class attribute: same for every Dog

    def __init__(self, name):
        self.name = name           # instance attribute: unique per Dog

a = Dog("Rex")
b = Dog("Bella")

print(a.species, b.species)   # both share the class attribute
print(a.name, b.name)         # each has its own name

Dog.species = "Updated species name"   # change on the CLASS
print(a.species, b.species)            # both change, because they share it

a.name = "Max"                          # change on ONE instance
print(a.name, b.name)                   # only a changes
```

### ⚠️ A common pitfall: mutable class attributes

If a class attribute is a mutable object (like a list), all instances share the *same* list
unless you're careful. This is a classic bug source.

```python
class Team:
    members = []   # ⚠️ shared list — BAD if you meant "each team has its own list"

    def add(self, name):
        self.members.append(name)

t1 = Team()
t2 = Team()
t1.add("Alice")
print(t1.members)   # ['Alice']
print(t2.members)   # ['Alice']  <-- surprising! They share the same list object

# Fix: initialize mutable attributes inside __init__ instead
class TeamFixed:
    def __init__(self):
        self.members = []   # a fresh list per instance

    def add(self, name):
        self.members.append(name)

t1 = TeamFixed()
t2 = TeamFixed()
t1.add("Alice")
print(t1.members, t2.members)
```

### Inspecting objects

Handy built-ins while learning/debugging:

```python
class Dog:
    def __init__(self, name):
        self.name = name

d = Dog("Rex")

print(type(d))          # the class of the object
print(isinstance(d, Dog))  # True
print(vars(d))           # dict of the object's instance attributes -> {'name': 'Rex'}
print(dir(d)[-3:])       # last few attributes/methods available (long list, so sliced)
```
