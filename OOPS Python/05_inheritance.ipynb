# 05 · Inheritance

**Goal:** learn how classes can reuse and extend other classes — single, multilevel,
hierarchical, and multiple inheritance — plus `super()` and the Method Resolution Order (MRO).

### What is inheritance?

Inheritance lets one class (the **child** / **subclass**) reuse the attributes and methods of
another class (the **parent** / **superclass**), and add or override behavior as needed.

Think: `Dog` and `Cat` are both `Animal`s — they share common behavior (`eat`, `sleep`) but
also have their own specific behavior (`bark`, `meow`).

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def sleep(self):
        return f"{self.name} is sleeping."


class Dog(Animal):          # Dog inherits from Animal
    def bark(self):
        return f"{self.name} says Woof!"


d = Dog("Rex")
print(d.eat())     # inherited from Animal
print(d.sleep())   # inherited from Animal
print(d.bark())    # defined in Dog itself
```

### `super()` — calling the parent's version

`super()` gives you access to the parent class, most commonly to extend `__init__` instead of
rewriting it from scratch.

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def make_sound(self):
        return f"{self.name} says {self.sound}"


class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, sound="Woof")   # reuse Animal's constructor
        self.species = "Canis familiaris"       # add Dog-specific attribute


d = Dog("Rex")
print(d.make_sound())
print(d.species)
```

### Method overriding

A child class can redefine (override) a parent method with its own implementation.

```python
class Animal:
    def make_sound(self):
        return "Some generic animal sound"


class Cat(Animal):
    def make_sound(self):                 # overrides Animal.make_sound
        return "Meow"


class Cow(Animal):
    def make_sound(self):
        base = super().make_sound()        # you can still call the parent's version if needed
        return f"{base}, well actually: Moo"


print(Cat().make_sound())
print(Cow().make_sound())
```

### Types of inheritance

Python supports every common inheritance pattern:

```python
# 1. Single inheritance: one parent, one child
class A: pass
class B(A): pass

# 2. Multilevel inheritance: a chain, grandparent -> parent -> child
class Animal:
    def breathe(self):
        return "breathing"

class Mammal(Animal):
    def feed_milk(self):
        return "feeding milk"

class Dog(Mammal):
    def bark(self):
        return "Woof"

d = Dog()
print(d.breathe())     # from Animal (2 levels up)
print(d.feed_milk())   # from Mammal (1 level up)
print(d.bark())        # from Dog itself

# 3. Hierarchical inheritance: multiple children from ONE parent
class Cat(Animal):
    def meow(self):
        return "Meow"

class Bird(Animal):
    def fly(self):
        return "Flying"

print(Cat().breathe(), Bird().breathe())   # both reuse Animal
```

```python
# 4. Multiple inheritance: a child inherits from MORE than one parent
class Swimmer:
    def swim(self):
        return "swimming"

class Flyer:
    def fly(self):
        return "flying"

class Duck(Swimmer, Flyer):   # Duck can do both!
    pass

duck = Duck()
print(duck.swim())
print(duck.fly())
```

### Method Resolution Order (MRO) — how Python picks *which* method to use

When multiple parents define the same method name, Python needs a rule for which one wins.
It uses the **C3 linearization algorithm**, but in practice: "left to right, depth first" is a
good enough mental model. You can always inspect it directly:

```python
class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):   # note the order: B before C
    pass

d = D()
print(d.greet())          # follows MRO -> finds B's version first
print(D.__mro__)          # shows the exact lookup order Python will use
```

### `isinstance()` vs `issubclass()`

Useful for checking relationships between objects and classes.

```python
class Animal: pass
class Dog(Animal): pass

d = Dog()

print(isinstance(d, Dog))      # True: d IS a Dog
print(isinstance(d, Animal))   # True: d IS ALSO an Animal (because of inheritance)
print(issubclass(Dog, Animal)) # True: Dog class inherits from Animal
print(issubclass(Animal, Dog)) # False: not the other way around
```
