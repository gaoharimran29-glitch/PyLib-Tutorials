# 06 · Polymorphism

**Goal:** understand "many forms" — how the same method name/operator can behave differently
depending on the object it's used with: method overriding, duck typing, and operator
overloading.

### What is polymorphism?

"Poly" = many, "morph" = forms. Polymorphism means the **same interface** (method name,
operator) works differently depending on the object it's applied to — without the calling
code needing to know exactly which type it's dealing with.

### 1. Polymorphism via method overriding (inheritance-based)

You already saw this in notebook 05 — different subclasses implement the same method name
differently, and calling code can treat them uniformly.

```python
class Shape:
    def area(self):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

shapes = [Rectangle(4, 5), Circle(3), Rectangle(2, 2)]

# Same loop, same method call (`.area()`), different behavior per object -- this IS polymorphism
for s in shapes:
    print(type(s).__name__, "->", s.area())
```

### 2. Duck typing — Python's flexible take on polymorphism

> "If it walks like a duck and quacks like a duck, it's a duck."

Python doesn't check an object's *type* before calling a method — it just tries to call it.
If the object has that method, it works, **regardless of inheritance**. This is more flexible
(and more Pythonic) than requiring a shared base class.

```python
class Duck:
    def sound(self):
        return "Quack!"

class Human:
    def sound(self):
        return "I'm imitating a duck: Quack!"

class Car:
    def sound(self):
        return "Honk! (not a duck, but it has a sound() method too)"

def make_it_sound(thing):
    # No type-checking here at all -- we just trust that `thing` has `.sound()`
    print(thing.sound())

for obj in [Duck(), Human(), Car()]:
    make_it_sound(obj)

# None of these classes share a common parent, yet the same function works on all of them.
```

### 3. Operator overloading — polymorphism for built-in operators

Python lets you define what `+`, `-`, `==`, `<`, `len()`, `str()` etc. mean for your own
objects, by implementing special "dunder" (double-underscore) methods. (Notebook 08 covers
these in depth — here's a taste.)

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):     # defines behavior for the `+` operator
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):            # defines what print()/str() shows
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2      # this calls v1.__add__(v2) automatically
print(v3)          # uses __str__ automatically
```

### The built-in `len()` is already polymorphic

A great everyday example: `len()` works differently depending on the type it receives, because
each type implements its own `__len__`.

```python
print(len("hello"))      # str.__len__
print(len([1, 2, 3]))    # list.__len__
print(len({"a": 1, "b": 2}))  # dict.__len__

class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.songs)

p = Playlist(["Song A", "Song B", "Song C"])
print(len(p))   # our own __len__ gets called too!
```

### Polymorphism with function arguments (built-in functions)

Functions like `+` also behave polymorphically on built-in types without any classes involved:

```python
print(2 + 3)          # numeric addition
print("Hi " + "there") # string concatenation
print([1, 2] + [3, 4]) # list concatenation

# All use the `+` operator, but the *behavior* depends on the type -- that's polymorphism.
```

2. What special method must a class define to support the `+` operator?
3. Why does `len(my_object)` work on custom classes?

<details>
<summary>Answers</summary>

1. Inheritance-based polymorphism requires a shared parent class with an overridden method;
   duck typing requires no shared parent at all — Python just checks the object has the method
   it's trying to call.
2. `__add__`.
3. Because `len()` internally calls the object's `__len__` method — if you define it, `len()`
   works on your custom class too.
