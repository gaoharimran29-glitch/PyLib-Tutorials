# 09 · Instance, Class, and Static Methods

**Goal:** clearly distinguish the three kinds of methods a Python class can have, and know
when to use each.

### The three kinds of methods

| Type | Decorator | First parameter | Can access | Typical use |
|---|---|---|---|---|
| Instance method | *(none)* | `self` | instance + class data | normal behavior tied to one object |
| Class method | `@classmethod` | `cls` | class data only | alternative constructors, factory methods |
| Static method | `@staticmethod` | *(neither)* | nothing automatic | utility function that's logically related to the class |

### 1. Instance methods (the default / most common)

These operate on a specific object and can access/modify that object's attributes via `self`.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def give_raise(self, amount):   # instance method -- needs self to know WHICH employee
        self.salary += amount

e = Employee("Asha", 50000)
e.give_raise(5000)
print(e.salary)
```

### 2. Class methods — operate on the class itself, not one instance

Marked with `@classmethod`, first parameter is `cls` (the class itself, not an object).
Very commonly used for **alternative constructors** — different ways to build an object.

```python
class Employee:
    company_name = "Acme Corp"     # class attribute
    employee_count = 0             # class attribute used as a shared counter

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.employee_count += 1   # every new Employee increments the shared counter

    @classmethod
    def from_string(cls, data_string):
        # Alternative constructor: build an Employee from "name-salary" text
        name, salary = data_string.split("-")
        return cls(name, int(salary))     # cls(...) is the same as calling Employee(...)

    @classmethod
    def change_company(cls, new_name):
        cls.company_name = new_name        # modifies the CLASS attribute, shared by all

e1 = Employee("Ravi", 40000)
e2 = Employee.from_string("Meera-55000")    # built via the alternative constructor

print(e2.name, e2.salary)
print(Employee.employee_count)               # 2 -- shared across all instances

Employee.change_company("Globex")
print(e1.company_name, e2.company_name)      # both updated -- it's a shared class attribute
```

### 3. Static methods — utility functions that live inside the class namespace

Marked with `@staticmethod`. They **don't** receive `self` or `cls` automatically — they behave
like a plain function, just namespaced inside the class because it's logically related.
Use these when the method doesn't need to touch instance or class data at all.

```python
class MathUtils:
    @staticmethod
    def is_even(n):
        return n % 2 == 0

    @staticmethod
    def add(a, b):
        return a + b

# Called on the class directly -- no object needed!
print(MathUtils.is_even(10))
print(MathUtils.add(3, 4))

# Also callable on an instance (works, but calling on the class is more common/clear)
m = MathUtils()
print(m.is_even(7))
```

### Putting all three together in one class

This mirrors a very common real-world pattern: instance methods for behavior, a classmethod
for a "smart" constructor, and a staticmethod for a related utility calculation.

```python
class Pizza:
    def __init__(self, toppings):
        self.toppings = toppings

    def __repr__(self):
        return f"Pizza({self.toppings})"

    # Instance method: depends on THIS pizza's toppings
    def add_topping(self, topping):
        self.toppings.append(topping)

    # Class method: alternative constructor ("factory")
    @classmethod
    def margherita(cls):
        return cls(["mozzarella", "tomato"])

    @classmethod
    def pepperoni(cls):
        return cls(["mozzarella", "tomato", "pepperoni"])

    # Static method: a related utility that needs no instance or class data
    @staticmethod
    def estimate_price(num_toppings):
        return 5 + num_toppings * 1.5

p1 = Pizza.margherita()
p2 = Pizza.pepperoni()
p1.add_topping("basil")

print(p1)
print(p2)
print(Pizza.estimate_price(len(p2.toppings)))
```

### 🧠 Quick check

1. What's the first parameter of a classmethod called, and what does it refer to?
2. Why would you use `@staticmethod` instead of just writing a normal function outside the
   class?
3. Give one common real-world use case for `@classmethod`.

<details>
<summary>Answers</summary>

1. `cls` — it refers to the class itself (e.g. `Employee`), not any particular instance.
2. Purely organizational — it groups a utility function with the class it logically belongs
   to, improving discoverability and namespacing (e.g. `MathUtils.is_even(x)`).
3. Alternative/"factory" constructors — e.g. building an object from a string, a dict, a file,
   or another format, without cluttering `__init__` with multiple unrelated signatures.
</details>
